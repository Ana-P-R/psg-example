## CORTOCIRCUITOS

```python
print ("Cortocircuito con operador and")
x = 1
y = 0
print (x > 2 and (x/y) > 2)
print (x > 0 and (x/y) > 0)

print ("Cortocircuito con operador or")
a = 1
b = 0
print (a > 0 or (a/b) > 0)
print (a > 2 or (a/b) > 2)
```
