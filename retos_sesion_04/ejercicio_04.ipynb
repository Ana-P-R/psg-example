# 🕹️ Batalla de Símbolos

Bienvenido a **Batalla de Símbolos**, un juego por turnos en el que dos jugadores se enfrentan en un tablero para conquistar la mayor cantidad de casillas usando símbolos únicos (x-o).


## 📜 Reglas del Juego

1. El juego se juega en un tablero de 5x5.
2. El jugador 1 usa el símbolo **`X`**, y el jugador 2 usa el símbolo **`O`**.
3. Los jugadores colocan su símbolo por turno en una celda vacía.
4. El objetivo es tener más símbolos en el tablero cuando esté completamente lleno.
5. El juego termina cuando todas las casillas están ocupadas.
